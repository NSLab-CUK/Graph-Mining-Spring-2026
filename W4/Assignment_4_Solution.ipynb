## Task 3: Vulnerability Analysis (Hub Pruning)

- **Data:** Re-load the `CE_PG` graph from [`data/CE-PG.txt`](data/CE-PG.txt).
- **Node Pruning (Attack Simulation):** Instead of removing unimportant nodes, simulate a targeted attack on the network by removing the **Top 10% most important nodes (hubs)** using the following centrality metrics:
    - Degree Centrality
    - Closeness Centrality
    - PageRank
- **Visualization:** For each pruning strategy, visualize the **remaining subgraph** and map the node sizes to their new **Closeness Centrality** scores to see how the graph fragmented.
- **Output:** Save the figures.

```python
# largest connected component
def get_largest_component(G):
    components = list(nx.connected_components(G))  
    if not components:
        raise ValueError("Graph has no nodes after filtering!")
    largest_component = max(components, key=len)
    return G.subgraph(largest_component).copy()

def draw_graph_closeness_centrality(largest_component, pos):
    # Calculate closeness centrality
    centrality = nx.closeness_centrality(largest_component)

    # Compute community structure
    lpc = nx.community.label_propagation_communities(largest_component)
    community_index = {n: i for i, com in enumerate(lpc) for n in com}

    # Draw graph
    fig, ax = plt.subplots(figsize=(20, 15))
    node_color = [community_index[n] for n in largest_component.nodes()]
    node_size = [v * 20000 for v in centrality.values()]

    nx.draw_networkx(
        largest_component,
        pos=pos,
        with_labels=False,
        node_color=node_color,
        node_size=node_size,
        edge_color="gainsboro",
        alpha=0.4,
    )

    # Title and legend
    font = {"color": "blue", "fontweight": "bold", "fontsize": 20}
    ax.set_title("Gene functional association network (C. elegans)", fontdict=font)
    
    font["color"] = "green"
    ax.text(0.80, 0.10, "Node color: community structure", 
            horizontalalignment="center", transform=ax.transAxes, fontdict=font)
    ax.text(0.80, 0.06, "Node size: closeness centrality", 
            horizontalalignment="center", transform=ax.transAxes, fontdict=font)

    ax.margins(0.1, 0.05)
    plt.axis("off")
    plt.show()
    
# Read graph data from edgelist
G = nx.read_weighted_edgelist("data/CE-PG.txt")

# Spring layout
largest_component = get_largest_component(G)
pos = nx.spring_layout(largest_component, k=0.15)
draw_graph_closeness_centrality(largest_component, pos)
```

```python
# Remove low-degree nodes
def remove_low_degree_nodes(G, min_degree=10):
    low_degree = [n for n, d in G.degree() if d < min_degree]
    G.remove_nodes_from(low_degree)
    return G
    
G1 = G.copy()
G1 = remove_low_degree_nodes(G1, min_degree=10)

print(f"Original nodes: {G.number_of_nodes()}")
print(f"Nodes after removing degree < 10: {G1.number_of_nodes()}")

# Get largest component
largest_component = get_largest_component(G1)

print(f"Largest component size: {largest_component.number_of_nodes()}")

# Layout + draw
pos = nx.spring_layout(largest_component, k=0.15, iterations=100)
draw_graph_closeness_centrality(largest_component, pos)
```

```python
def remove_low_closeness_nodes(G, threshold=0.3):
    G = G.copy()
    
    try:
        closeness_centralities = nx.closeness_centrality(G)
    except nx.NetworkXError:
        print("Error: Graph is empty or not connected enough for closeness centrality.")
        return G
    
    low_closeness_nodes = [
        node for node, closeness in closeness_centralities.items() 
        if closeness < threshold
    ]
    
    G.remove_nodes_from(low_closeness_nodes)
    return G

G2 = G.copy()                               
G2 = remove_low_closeness_nodes(G2, threshold=0.3)

print(f"Original nodes: {G.number_of_nodes()}")
print(f"Nodes after removing low closeness (< 0.3): {G2.number_of_nodes()}")

# Get largest connected component
largest_component = get_largest_component(G2)

print(f"Largest component size: {largest_component.number_of_nodes()} nodes")

# Draw
pos = nx.spring_layout(largest_component, k=0.15, iterations=100)
draw_graph_closeness_centrality(largest_component, pos)
```

```python
def remove_low_pagerank_nodes(G, threshold=0.005): 
    G = G.copy()
    
    try:
        pagerank_centralities = nx.pagerank(G, alpha=0.85)
    except Exception as e:
        print(f"Error calculating PageRank: {e}")
        return G
    
    low_pagerank_nodes = [
        node for node, pr in pagerank_centralities.items() 
        if pr < threshold
    ]
    
    G.remove_nodes_from(low_pagerank_nodes)
    return G



G4 = G.copy()
G4 = remove_low_pagerank_nodes(G4, threshold=0.0005)   

print(f"Original nodes: {G.number_of_nodes()}")
print(f"Nodes after removing low PageRank: {G4.number_of_nodes()}")

# Get largest connected component
largest_component = get_largest_component(G4)

print(f"Largest component size: {largest_component.number_of_nodes()} nodes")

# Draw
pos = nx.spring_layout(largest_component, k=0.15, iterations=100)
draw_graph_closeness_centrality(largest_component, pos)
```
