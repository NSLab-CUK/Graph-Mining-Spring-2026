# Additional question
Write a Python function that takes a graph G and a minimum support count as input and returns the top-k most frequent subgraphs in G. The function should use a combination of the Apriori algorithm and a priority queue to efficiently mine and rank frequent subgraphs


G = {
    'A': ['B', 'C', 'D'],
    'B': ['A', 'C'],
    'C': ['A', 'B', 'D'],
    'D': ['A', 'C']
}

min_support = 2

k = 3

```python
import heapq
from collections import defaultdict
import networkx as nx
from itertools import combinations

def graph_to_nx(G_dict):
    G = nx.Graph()
    for u, neighbors in G_dict.items():
        for v in neighbors:
            if u < v:  # undirected, add once
                G.add_edge(u, v)
    return G

def get_canonical_form(subg):
    """Return a canonical string representation of the subgraph for deduping (unlabeled structure)."""
    nodes = sorted(subg.nodes())
    edges = sorted((min(u,v), max(u,v)) for u,v in subg.edges())
    return str(nodes) + '|' + str(edges)

def count_embeddings(G, pattern):
    """Count number of subgraph isomorphisms from pattern to G (support). Simple matcher for small G."""
    matcher = nx.algorithms.isomorphism.GraphMatcher(G, pattern)
    return sum(1 for _ in matcher.subgraph_isomorphisms_iter())

def find_top_k_frequent_subgraphs(G_dict, min_support, k):
    """
    Mine top-k frequent subgraphs from single graph G using Apriori-style extension + max-heap (priority queue).
    Patterns are treated as unlabeled (structural). Support = # of subgraph embeddings.
    Returns list of (support, subgraph_edge_list, canonical) sorted by support desc.
    """
    G = graph_to_nx(G_dict)
    # Level-wise: start with frequent edges (size 2)
    # Use priority queue (max-heap via negative support) for top-k overall
    topk_heap = []  # min-heap of (-sup, canon, edges)
    seen = set()
    
    # 1-node patterns (trivial)
    node_pat = nx.Graph()
    node_pat.add_node(0)
    sup = G.number_of_nodes()
    if sup >= min_support:
        canon = get_canonical_form(node_pat)
        seen.add(canon)
        heapq.heappush(topk_heap, (-sup, canon, []))
    
    # Frequent edges as seeds (Apriori level 1)
    edge_patterns = []
    for u, v in G.edges():
        pat = nx.Graph()
        pat.add_edge(0, 1)
        sup = count_embeddings(G, pat)
        if sup >= min_support:
            canon = get_canonical_form(pat)
            if canon not in seen:
                seen.add(canon)
                heapq.heappush(topk_heap, (-sup, canon, [(0,1)]))
                edge_patterns.append(pat)
    
    # Apriori-style: extend current frequent patterns by adding a node + edge (grow by one vertex)
    # For small graph we do limited extensions (BFS/DFS style candidate gen)
    current_level = [p for p in edge_patterns]
    max_size = G.number_of_nodes()
    
    for size in range(3, max_size+1):
        next_level = []
        for pat in current_level:
            # Try to extend pat by adding a new vertex connected to one existing
            num_nodes = pat.number_of_nodes()
            for attach_to in list(pat.nodes()):
                # add new node
                new_pat = pat.copy()
                new_v = num_nodes
                new_pat.add_edge(attach_to, new_v)
                # check support
                sup = count_embeddings(G, new_pat)
                canon = get_canonical_form(new_pat)
                if sup >= min_support and canon not in seen:
                    seen.add(canon)
                    heapq.heappush(topk_heap, (-sup, canon, list(new_pat.edges())))
                    next_level.append(new_pat)
                    # keep only top k in heap to prune (optional, here keep all then trim)
        current_level = next_level
        if not current_level:
            break
    
    # Extract top k (highest support). Use heap to pop largest
    results = []
    while topk_heap and len(results) < k:
        negsup, canon, edgelist = heapq.heappop(topk_heap)
        sup = -negsup
        results.append((sup, edgelist, canon))
    
    # sort just in case by sup desc
    results.sort(key=lambda x: -x[0])
    return results[:k]

# Given input
G = {
    'A': ['B', 'C', 'D'],
    'B': ['A', 'C'],
    'C': ['A', 'B', 'D'],
    'D': ['A', 'C']
}
min_support = 2
k = 3

topk = find_top_k_frequent_subgraphs(G, min_support, k)
print("Top-3 most frequent subgraphs (support, edges_in_pattern, canon):")
for rank, (sup, edges, canon) in enumerate(topk, 1):
    print(f"{rank}. support={sup}, edges={edges}, canon={canon}")

# For interpretability also print the actual patterns as networkx info
print("\nNote: Supports computed via subgraph isomorphism count (unlabeled patterns).")
```

# Conceptual question
What is the difference between graph isomorphism and subgraph isomorphism?

Graph Isomorphism:
- Two graphs G1 and G2 are isomorphic if there exists a bijective (1-to-1 and onto) mapping 
  between their vertices such that edges are preserved exactly (adjacency is identical after relabeling).
- They have exactly the same structure; same number of nodes, edges, degrees, etc.
- It is a special case of subgraph isomorphism where the mapping covers all vertices of both graphs.
- Example: any two cycle graphs are isomorphic.

Subgraph Isomorphism:
- A (smaller) graph H is subgraph-isomorphic to G (or G contains H as subgraph) if there exists 
  an injective mapping from vertices of H into vertices of G such that edges in H are preserved 
  in G (but G may have extra vertices/edges).
- Not necessarily bijective/surjective onto G; G can be larger and have additional structure.
- Used heavily in frequent subgraph mining (finding patterns that embed many times).
- Example: a triangle is subgraph-isomorphic, but not isomorphic (different sizes).

Key Difference: Isomorphism requires perfect structure match (same size), subgraph isomorphism 
allows the pattern to be "contained inside" a larger host graph.
